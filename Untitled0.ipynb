```python
import numpy as np, pandas as pd
np.random.seed(42)

def simulate(n=3000):
    tank, bottle, valve = 80.0, 0.0, 1
    rows = []
    for t in range(n):
        if bottle >= 95: bottle, tank, valve = 0.0, 80.0, 1
        flow = 2.0 if valve else 0.0
        tank -= flow*0.5; bottle += flow*0.5
        rows.append([t*0.5, tank+np.random.normal(0,.1),
                     flow+np.random.normal(0,.05),
                     bottle+np.random.normal(0,.1), valve])
    return pd.DataFrame(rows, columns=["time","tank","flow","bottle","valve"])

df = simulate(); df["label"] = "Normal"

def attack(df, kind, start, end):
    idx = df.index[start:end]
    if kind == "NMM":   # naive: constant/random
        df.loc[idx,"flow"] = np.random.uniform(0,3,len(idx))
    elif kind == "CMM": # calculated: (1 +- factor) * value
        df.loc[idx,"bottle"] *= (1 - 0.15)
    elif kind == "CI":  # command injection: valve toggle
        df.loc[idx,"valve"] = [i%2 for i in range(len(idx))]
    df.loc[idx,"label"] = kind

for k,(s,e) in {"NMM":(300,600),"CMM":(900,1300),
                "CI":(1600,1800),"NMM ":(2100,2300),"CMM ":(2500,2800)}.items():
    attack(df, k.strip(), s, e)

df["flow_x_valve"] = df["flow"] * df["valve"]
df.to_csv("dataset.csv", index=False)
print(df.label.value_counts())
```

```python
import time, warnings
warnings.filterwarnings("ignore")
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.dummy import DummyClassifier
from sklearn.metrics import classification_report, confusion_matrix

X = df[["tank","flow","bottle","valve","flow_x_valve"]]
y = df["label"]
Xtr,Xte,ytr,yte = train_test_split(X,y,test_size=0.3,stratify=y,random_state=42)

models = {"Baseline":DummyClassifier(strategy="most_frequent"),
          "LogReg":LogisticRegression(max_iter=1000),
          "DecisionTree":DecisionTreeClassifier(random_state=42),
          "RandomForest":RandomForestClassifier(random_state=42)}
for name,m in models.items():
    m.fit(Xtr,ytr)
    t=time.time(); p=m.predict(Xte); dt=(time.time()-t)/len(Xte)*1000
    print("=====",name,f"| {dt:.4f} ms/sample")
    print(classification_report(yte,p,zero_division=0))
    print(confusion_matrix(yte,p))
```

```python
import warnings, os
warnings.filterwarnings("ignore")
os.environ["PYTHONWARNINGS"] = "ignore"
import multiprocessing as mp, psutil, time

def producer(q):          # alag process: simulator
    for _,r in df.sample(200, random_state=1).iterrows():
        q.put(r[["tank","flow","bottle","valve","flow_x_valve"]].values)
    q.put(None)

def detector(q, lock, log):   # alag process: IDS
    import warnings; warnings.filterwarnings("ignore")
    rf = models["RandomForest"]; p = psutil.Process(os.getpid())
    while True:
        x = q.get()
        if x is None: break
        pred = rf.predict(pd.DataFrame([x], columns=["tank","flow","bottle","valve","flow_x_valve"]))[0]
        with lock:                # synchronization
            log.append((pred, p.cpu_percent(), p.memory_info().rss/1e6))

q = mp.Queue(); lock = mp.Lock(); log = mp.Manager().list()
a = mp.Process(target=producer,args=(q,))
b = mp.Process(target=detector,args=(q,lock,log))
a.start(); b.start(); a.join(); b.join()
out = pd.DataFrame(list(log),columns=["pred","cpu%","mem_MB"])
out.to_csv("os_log.csv",index=False)
print(out.describe())
```

```python
import warnings; warnings.filterwarnings("ignore")
from sklearn.metrics import f1_score, accuracy_score
rng = np.random.default_rng(42)

# simulated network features (report mein likhna: "simulated")
df["pkt_rate"] = rng.normal(50,3,len(df))
df["resp_ms"]  = rng.normal(5,0.5,len(df))
df.loc[df.label=="CI","pkt_rate"] += 30
df.loc[df.label.isin(["NMM","CMM"]),"pkt_rate"] += 8
df.loc[df.label.isin(["NMM","CMM"]),"resp_ms"] += 2

groups = {
 "process_only": ["tank","flow","bottle"],
 "process+control": ["tank","flow","bottle","valve","flow_x_valve"],
 "process+control+network": ["tank","flow","bottle","valve","flow_x_valve","pkt_rate","resp_ms"]}
res=[]
for g,cols in groups.items():
    a,b,c,d = train_test_split(df[cols],df["label"],test_size=0.3,stratify=df["label"],random_state=42)
    p = RandomForestClassifier(random_state=42).fit(a,c).predict(b)
    res.append([g, round(accuracy_score(d,p),3), round(f1_score(d,p,average="macro"),3)])
E4 = pd.DataFrame(res, columns=["features","accuracy","macro_F1"])
print(E4)
```

```python
proc = psutil.Process(os.getpid())
cols5 = ["tank","flow","bottle","valve","flow_x_valve"]
Xtr,Xte,ytr,yte = train_test_split(df[cols5],df["label"],test_size=0.3,stratify=df["label"],random_state=42)
rows=[]
for name,m in models.items():
    mem0 = proc.memory_info().rss/1e6; c0 = proc.cpu_times()
    t=time.time(); m.fit(Xtr,ytr); fit_t=time.time()-t
    c1 = proc.cpu_times()
    t=time.time(); m.predict(Xte); inf=(time.time()-t)/len(Xte)*1000
    rows.append([name, round(fit_t,3), round(c1.user-c0.user,3),
                 round(proc.memory_info().rss/1e6-mem0,1), round(inf,4)])
E5 = pd.DataFrame(rows, columns=["model","train_s","train_cpu_s","mem_delta_MB","infer_ms_per_sample"])
print(E5)
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay
rng = np.random.default_rng(42)
rf = RandomForestClassifier(random_state=42).fit(Xtr,ytr)
sd = Xte.std().values
out=[]
for k in [0,0.1,0.25,0.5,1.0]:
    Xn = Xte + rng.normal(0,1,Xte.shape)*sd*k
    out.append([k, round(f1_score(yte,rf.predict(Xn),average="macro"),3)])
E6 = pd.DataFrame(out, columns=["noise_level(x std)","macro_F1"]); print(E6)

plt.figure(); plt.plot(E6.iloc[:,0],E6.iloc[:,1],marker="o")
plt.xlabel("Noise level (x std)"); plt.ylabel("Macro F1"); plt.title("E6: Robustness")
plt.savefig("e6_noise.png",dpi=150)

ConfusionMatrixDisplay.from_estimator(rf,Xte,yte,cmap="Blues")
plt.title("Random Forest confusion matrix"); plt.savefig("cm_rf.png",dpi=150)

E4.plot.bar(x="features",y="macro_F1",legend=False); plt.title("E4: Feature groups")
plt.tight_layout(); plt.savefig("e4_features.png",dpi=150)
```
